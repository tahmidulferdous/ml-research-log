```python
from diffusers import DiffusionPipeline
import torch

# Load the pipeline
pipe = DiffusionPipeline.from_pretrained("Wan-AI/Wan2.1-T2V-14B", torch_dtype=torch.float16)
pipe = pipe.to("cuda") # Move to GPU

# Generate video from text
prompt = "A cat playing with a ball of yarn"
video_frames = pipe(prompt, num_inference_steps=50).frames

# Save as video (requires additional libraries like imageio)
import imageio
imageio.mimsave("generated_video.mp4", video_frames, fps=8)
```
