```python
from transformers import AutoTokenizer
tok = AutoTokenizer.from_pretrained("shahidul034/BanglaGPT")
print("tokenizer type:", type(tok))
# quick coverage check (token length / avg tokens per sentence)
texts = ["বিয়াক্ মানুশ ইজ্‌জত এদ্দে অ়কর্ ই়শাফে আজাদ্ আর্ ও়ঁৰাইন্‌না অ়ইয়েরে ফ়য়দা অ়য়্। ই়তারাত্‌তু আহল্ এদ্‌দে বিবেক্ আস়ে; এ়তল্‌লায়্ এজ্‌জন্ আরেজ্‌জনর্ ও়ঁৰারে ৱেইয়র না৴ন বেৱার গরন দরহারি।"]  # load a sample of your corpus
lens = [len(tok(t).input_ids) for t in texts]
print("avg tokens:", sum(lens)/len(lens), "max:", max(lens))
```
