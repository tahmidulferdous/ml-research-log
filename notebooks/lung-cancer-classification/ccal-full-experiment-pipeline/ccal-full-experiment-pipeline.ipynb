```python

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, random_split
import torchvision.transforms as transforms
from torchvision.transforms import functional as TF
import numpy as np
import pandas as pd
from PIL import Image
import cv2
import os
from pathlib import Path
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix
from sklearn.model_selection import StratifiedKFold
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')

# ==================== Global Configuration ====================
class Config:
    """Global configuration for the CCAL model"""
    # Paths
    DATA_PATH = "/kaggle/input/the-iqothnccd-lung-cancer-dataset/The IQ-OTHNCCD lung cancer dataset"  # Update with your dataset path
    SAVE_PATH = "/saved_models"
    
    # Image settings
    IMAGE_SIZE = 224
    CHANNELS = 3
    
    # Training parameters
    BATCH_SIZE = 32
    EPOCHS = 50
    LEARNING_RATE = 1e-5
    DROPOUT = 0.5
    TRAIN_SPLIT = 0.8
    NUM_CLASSES = 3
    
    # Model parameters
    MHA_HEADS = 4
    MHA_KEY_DIM = 16
    MHA_VALUE_DIM = 64
    MLP_UNITS = 256
    
    # Device
    DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    
    # Class mapping
    CLASS_NAMES = ['Bengin cases', 'Malignant cases', 'Normal cases']
    CLASS_MAPPING = {'Bengin cases': 0, 'Malignant cases': 1, 'Normal cases': 2}

# ==================== Data Preprocessing ====================
class CLAHETransform:
    """Apply CLAHE (Contrast Limited Adaptive Histogram Equalization) to images"""
    def __init__(self, clip_limit=2.0, tile_grid_size=(8, 8)):
        self.clahe = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=tile_grid_size)
    
    def __call__(self, img):
        if isinstance(img, Image.Image):
            img = np.array(img)
        
        if len(img.shape) == 3:
            # Convert to LAB color space
            lab = cv2.cvtColor(img, cv2.COLOR_RGB2LAB)
            l, a, b = cv2.split(lab)
            
            # Apply CLAHE to L channel
            l_clahe = self.clahe.apply(l)
            
            # Merge channels
            lab_clahe = cv2.merge((l_clahe, a, b))
            
            # Convert back to RGB
            img_clahe = cv2.cvtColor(lab_clahe, cv2.COLOR_LAB2RGB)
        else:
            img_clahe = self.clahe.apply(img)
        
        return Image.fromarray(img_clahe)
        

class LungCancerDataset(Dataset):
    """Custom dataset for lung cancer CT images"""
    def __init__(self, root_dir, transform=None, train=True, train_split=0.8):
        self.root_dir = Path(root_dir)
        self.transform = transform
        self.train = train
        
        # Load all images and labels
        self.images = []
        self.labels = []
        
        for class_idx, class_name in enumerate(Config.CLASS_NAMES):
            class_dir = self.root_dir / class_name
            if class_dir.exists():
                for img_path in class_dir.glob('*.jpg'):  # Adjust extension as needed
                    self.images.append(str(img_path))
                    self.labels.append(class_idx)
        
        # Convert to numpy arrays for splitting
        self.images = np.array(self.images)
        self.labels = np.array(self.labels)
        
        # Train/test split
        np.random.seed(42)
        indices = np.arange(len(self.images))
        np.random.shuffle(indices)
        
        split_idx = int(len(indices) * train_split)
        if train:
            indices = indices[:split_idx]
        else:
            indices = indices[split_idx:]
        
        self.images = self.images[indices]
        self.labels = self.labels[indices]
    
    def __len__(self):
        return len(self.images)
    
    def __getitem__(self, idx):
        img_path = self.images[idx]
        label = self.labels[idx]
        
        # Load image
        image = Image.open(img_path).convert('RGB')
        
        # Apply transformations
        if self.transform:
            image = self.transform(image)
        
        return image, label

def get_data_transforms():
    """Get data transformation pipelines"""
    train_transform = transforms.Compose([
        CLAHETransform(),
        transforms.Resize((Config.IMAGE_SIZE, Config.IMAGE_SIZE)),
        transforms.RandomRotation(degrees=10),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomAffine(degrees=0, translate=(0.1, 0.1)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], 
                           std=[0.229, 0.224, 0.225])
    ])
    
    test_transform = transforms.Compose([
        CLAHETransform(),
        transforms.Resize((Config.IMAGE_SIZE, Config.IMAGE_SIZE)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], 
                           std=[0.229, 0.224, 0.225])
    ])
    
    return train_transform, test_transform

# ==================== Model Architecture ====================
class CNN_Model1(nn.Module):
    """First CNN branch with 4 convolutional layers"""
    def __init__(self):
        super(CNN_Model1, self).__init__()
        
        self.conv_blocks = nn.Sequential(
            # Block 1
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
            
            # Block 2
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
            
            # Block 3
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(128, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
            
            # Block 4
            nn.Conv2d(128, 256, kernel_size=3, padding=1),
            nn.ReLU(),
        )
        
        self.global_pool = nn.AdaptiveMaxPool2d(1)
    
    def forward(self, x):
        x = self.conv_blocks(x)
        x = self.global_pool(x)
        x = x.view(x.size(0), -1)
        return x

class CNN_Model2(nn.Module):
    """Second CNN branch with 2 convolutional layers"""
    def __init__(self):
        super(CNN_Model2, self).__init__()
        
        self.conv_blocks = nn.Sequential(
            # Block 1
            nn.Conv2d(3, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
            
            # Block 2
            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
        )
        
        self.global_pool = nn.AdaptiveMaxPool2d(1)
    
    def forward(self, x):
        x = self.conv_blocks(x)
        x = self.global_pool(x)
        x = x.view(x.size(0), -1)
        return x

class MultiHeadAttention(nn.Module):
    """Multi-Head Attention mechanism"""
    def __init__(self, embed_dim, num_heads=4, key_dim=16, value_dim=64):
        super(MultiHeadAttention, self).__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.key_dim = key_dim
        self.value_dim = value_dim
        
        # Linear projections for Q, K, V
        self.W_q = nn.Linear(embed_dim, num_heads * key_dim)
        self.W_k = nn.Linear(embed_dim, num_heads * key_dim)
        self.W_v = nn.Linear(embed_dim, num_heads * value_dim)
        self.W_o = nn.Linear(num_heads * value_dim, embed_dim)
        
        self.scale = key_dim ** -0.5
    
    def forward(self, x):
        batch_size, seq_len, _ = x.shape
        
        # Generate Q, K, V
        Q = self.W_q(x).view(batch_size, seq_len, self.num_heads, self.key_dim)
        K = self.W_k(x).view(batch_size, seq_len, self.num_heads, self.key_dim)
        V = self.W_v(x).view(batch_size, seq_len, self.num_heads, self.value_dim)
        
        # Transpose for attention computation
        Q = Q.transpose(1, 2)  # [batch, heads, seq_len, key_dim]
        K = K.transpose(1, 2)
        V = V.transpose(1, 2)
        
        # Compute attention scores
        scores = torch.matmul(Q, K.transpose(-2, -1)) * self.scale
        attention_weights = F.softmax(scores, dim=-1)
        
        # Apply attention to values
        attention_output = torch.matmul(attention_weights, V)
        
        # Concatenate heads
        attention_output = attention_output.transpose(1, 2).contiguous()
        attention_output = attention_output.view(batch_size, seq_len, -1)
        
        # Final projection
        output = self.W_o(attention_output)
        
        return output

class GELU(nn.Module):
    """Gaussian Error Linear Unit activation"""
    def forward(self, x):
        return 0.5 * x * (1 + torch.tanh(np.sqrt(2 / np.pi) * (x + 0.044715 * torch.pow(x, 3))))

class MLPBlock(nn.Module):
    """MLP block with GELU activation"""
    def __init__(self, input_dim, hidden_dim=256, num_classes=3,dropout_p = Config.DROPOUT):
        super(MLPBlock, self).__init__()
        
        self.global_avg_pool = nn.AdaptiveAvgPool1d(1)
        self.batch_norm1 = nn.BatchNorm1d(input_dim)
        self.dense1 = nn.Linear(input_dim, hidden_dim)
        self.gelu = GELU()
        self.batch_norm2 = nn.BatchNorm1d(hidden_dim)
        self.dropout = nn.Dropout(dropout_p)
        self.dense2 = nn.Linear(hidden_dim, num_classes)
    
    def forward(self, x):
        # x shape: [batch, seq_len, features]
        x = x.transpose(1, 2)  # [batch, features, seq_len]
        x = self.global_avg_pool(x).squeeze(-1)  # [batch, features]
        
        x = self.batch_norm1(x)
        x = self.dense1(x)
        x = self.gelu(x)
        x = self.batch_norm2(x)
        x  = self.dropout(x)
        x = self.dense2(x)
        
        return x

class CCAL_Model(nn.Module):
    """Complete CCAL (Concatenated CNN Attention Lung) model"""
    def __init__(self):
        super(CCAL_Model, self).__init__()
        
        # Two CNN branches
        self.cnn1 = CNN_Model1()
        self.cnn2 = CNN_Model2()
        
        # Calculate concatenated dimension
        self.concat_dim = 256 + 32  # CNN1 output: 256, CNN2 output: 32
        
        # Multi-Head Attention layers
        self.mha1 = MultiHeadAttention(
            embed_dim=self.concat_dim,
            num_heads=Config.MHA_HEADS,
            key_dim=Config.MHA_KEY_DIM,
            value_dim=Config.MHA_VALUE_DIM
        )
        self.mha2 = MultiHeadAttention(
            embed_dim=self.concat_dim,
            num_heads=Config.MHA_HEADS,
            key_dim=Config.MHA_KEY_DIM,
            value_dim=Config.MHA_VALUE_DIM
        )
        
        # MLP block
        self.mlp = MLPBlock(
            input_dim=self.concat_dim,
            hidden_dim=Config.MLP_UNITS,
            num_classes=Config.NUM_CLASSES
        )
    
    def forward(self, x):
        # Process through both CNN branches
        features1 = self.cnn1(x)
        features2 = self.cnn2(x)
        
        # Concatenate features
        combined = torch.cat([features1, features2], dim=1)
        
        # Reshape for attention layers
        combined = combined.unsqueeze(1)  # Add sequence dimension
        
        # Apply Multi-Head Attention
        attended = self.mha1(combined)
        attended = self.mha2(attended)
        
        # Apply MLP block
        output = self.mlp(attended)
        
        return output

# ==================== Training and Evaluation ====================
class Trainer:
    """Training and evaluation handler for CCAL model"""
    def __init__(self, model, device=Config.DEVICE):
        self.model = model.to(device)
        self.device = device
        self.criterion = nn.CrossEntropyLoss()
        self.optimizer = torch.optim.Adam(model.parameters(), lr=Config.LEARNING_RATE)
        self.history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
    
    def train_epoch(self, dataloader):
        self.model.train()
        total_loss = 0
        correct = 0
        total = 0
        
        for batch_idx, (inputs, targets) in enumerate(tqdm(dataloader, desc="Training")):
            inputs, targets = inputs.to(self.device), targets.to(self.device)
            
            self.optimizer.zero_grad()
            outputs = self.model(inputs)
            loss = self.criterion(outputs, targets)
            
            loss.backward()
            self.optimizer.step()
            
            total_loss += loss.item()
            _, predicted = outputs.max(1)
            total += targets.size(0)
            correct += predicted.eq(targets).sum().item()
        
        avg_loss = total_loss / len(dataloader)
        accuracy = 100. * correct / total
        
        return avg_loss, accuracy
    
    def validate(self, dataloader):
        self.model.eval()
        total_loss = 0
        correct = 0
        total = 0
        
        all_preds = []
        all_targets = []
        all_probs = []
        
        with torch.no_grad():
            for batch_idx, (inputs, targets) in enumerate(tqdm(dataloader, desc="Validation")):
                inputs, targets = inputs.to(self.device), targets.to(self.device)
                
                outputs = self.model(inputs)
                loss = self.criterion(outputs, targets)
                
                total_loss += loss.item()
                _, predicted = outputs.max(1)
                total += targets.size(0)
                correct += predicted.eq(targets).sum().item()
                
                # Store for metrics calculation
                probs = F.softmax(outputs, dim=1)
                all_probs.extend(probs.cpu().numpy())
                all_preds.extend(predicted.cpu().numpy())
                all_targets.extend(targets.cpu().numpy())
        
        avg_loss = total_loss / len(dataloader)
        accuracy = 100. * correct / total
        
        # Calculate additional metrics
        metrics = self.calculate_metrics(all_targets, all_preds, all_probs)
        
        return avg_loss, accuracy, metrics
    
    def calculate_metrics(self, y_true, y_pred, y_probs):
        """Calculate comprehensive metrics"""
        metrics = {}
        
        # Basic metrics
        metrics['accuracy'] = accuracy_score(y_true, y_pred)
        metrics['precision'] = precision_score(y_true, y_pred, average='weighted')
        metrics['recall'] = recall_score(y_true, y_pred, average='weighted')
        metrics['f1'] = f1_score(y_true, y_pred, average='weighted')
        
        # AUC for multi-class
        y_probs = np.array(y_probs)
        if Config.NUM_CLASSES > 2:
            metrics['auc'] = roc_auc_score(y_true, y_probs, multi_class='ovr', average='weighted')
        else:
            metrics['auc'] = roc_auc_score(y_true, y_probs[:, 1])
        
        # Confusion matrix
        metrics['confusion_matrix'] = confusion_matrix(y_true, y_pred)
        
        # Per-class metrics
        for i in range(Config.NUM_CLASSES):
            class_name = Config.CLASS_NAMES[i]
            y_true_binary = (np.array(y_true) == i).astype(int)
            y_pred_binary = (np.array(y_pred) == i).astype(int)
            
            metrics[f'{class_name}_precision'] = precision_score(y_true_binary, y_pred_binary)
            metrics[f'{class_name}_recall'] = recall_score(y_true_binary, y_pred_binary)
            metrics[f'{class_name}_f1'] = f1_score(y_true_binary, y_pred_binary)
        
        return metrics
    
    def train(self, train_loader, val_loader, epochs=Config.EPOCHS):
        """Full training loop"""
        best_acc = 0
        
        for epoch in range(epochs):
            print(f"\nEpoch {epoch+1}/{epochs}")
            print("-" * 50)
            
            # Training
            train_loss, train_acc = self.train_epoch(train_loader)
            
            # Validation
            val_loss, val_acc, metrics = self.validate(val_loader)
            
            # Store history
            self.history['train_loss'].append(train_loss)
            self.history['train_acc'].append(train_acc)
            self.history['val_loss'].append(val_loss)
            self.history['val_acc'].append(val_acc)
            
            # Print metrics
            print(f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%")
            print(f"Val Loss: {val_loss:.4f} | Val Acc: {val_acc:.2f}%")
            print(f"Precision: {metrics['precision']:.4f} | Recall: {metrics['recall']:.4f}")
            print(f"F1-Score: {metrics['f1']:.4f} | AUC: {metrics['auc']:.4f}")
            
            # Save best model
            if val_acc > best_acc:
                best_acc = val_acc
                self.save_model(f"best_model_acc_{val_acc:.2f}.pth")
                print(f"Saved best model with accuracy: {val_acc:.2f}%")
        
        return metrics
    
    def save_model(self, filename):
        """Save model checkpoint"""
        os.makedirs(Config.SAVE_PATH, exist_ok=True)
        filepath = os.path.join(Config.SAVE_PATH, filename)
        torch.save({
            'model_state_dict': self.model.state_dict(),
            'optimizer_state_dict': self.optimizer.state_dict(),
            'history': self.history
        }, filepath)
    
    def load_model(self, filepath):
        """Load model checkpoint"""
        checkpoint = torch.load(filepath)
        self.model.load_state_dict(checkpoint['model_state_dict'])
        self.optimizer.load_state_dict(checkpoint['optimizer_state_dict'])
        self.history = checkpoint['history']

def plot_training_history(history):
    """Plot training and validation metrics"""
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    
    # Loss plot
    axes[0].plot(history['train_loss'], label='Train Loss')
    axes[0].plot(history['val_loss'], label='Val Loss')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Loss')
    axes[0].set_title('Training and Validation Loss')
    axes[0].legend()
    axes[0].grid(True)
    
    # Accuracy plot
    axes[1].plot(history['train_acc'], label='Train Accuracy')
    axes[1].plot(history['val_acc'], label='Val Accuracy')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Accuracy (%)')
    axes[1].set_title('Training and Validation Accuracy')
    axes[1].legend()
    axes[1].grid(True)
    
    plt.tight_layout()
    plt.savefig('training_history.png')
    plt.show()

def plot_confusion_matrix(cm, class_names):
    """Plot confusion matrix"""
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=class_names, yticklabels=class_names)
    plt.title('Confusion Matrix')
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()
    plt.savefig('confusion_matrix.png')
    plt.show()

def k_fold_cross_validation(dataset, k=5):
    """Perform k-fold cross-validation"""
    kfold = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)
    
    fold_results = []
    
    for fold, (train_idx, val_idx) in enumerate(kfold.split(dataset.images, dataset.labels)):
        print(f"\nFold {fold+1}/{k}")
        print("-" * 50)
        
        # Create data subsets
        train_subset = torch.utils.data.Subset(dataset, train_idx)
        val_subset = torch.utils.data.Subset(dataset, val_idx)
        
        # Create dataloaders
        train_loader = DataLoader(train_subset, batch_size=Config.BATCH_SIZE, 
                                 shuffle=True, num_workers=4)
        val_loader = DataLoader(val_subset, batch_size=Config.BATCH_SIZE, 
                               shuffle=False, num_workers=4)
        
        # Initialize model and trainer
        model = CCAL_Model()
        trainer = Trainer(model)
        
        # Train for a few epochs (reduce for CV)
        metrics = trainer.train(train_loader, val_loader, epochs=10)
        
        fold_results.append({
            'fold': fold + 1,
            'accuracy': metrics['accuracy'],
            'precision': metrics['precision'],
            'recall': metrics['recall'],
            'f1': metrics['f1'],
            'auc': metrics['auc']
        })
    
    # Calculate average metrics
    avg_metrics = {}
    for key in ['accuracy', 'precision', 'recall', 'f1', 'auc']:
        values = [r[key] for r in fold_results]
        avg_metrics[key] = np.mean(values)
        avg_metrics[f'{key}_std'] = np.std(values)
    
    return fold_results, avg_metrics

# ==================== Main Execution ====================
def main():
    """Main execution function"""
    print("=" * 60)
    print("CCAL Model - Lung Cancer Classification")
    print("=" * 60)
    
    # Set random seeds for reproducibility
    torch.manual_seed(42)
    np.random.seed(42)
    
    print(f"\nUsing device: {Config.DEVICE}")
    
    # 1. Data Preparation
    print("\n1. Loading and preparing data...")
    train_transform, test_transform = get_data_transforms()
    
    # Create datasets
    train_dataset = LungCancerDataset(
        Config.DATA_PATH, 
        transform=train_transform, 
        train=True
    )
    test_dataset = LungCancerDataset(
        Config.DATA_PATH, 
        transform=test_transform, 
        train=False
    )
    
    print(f"Train samples: {len(train_dataset)}")
    print(f"Test samples: {len(test_dataset)}")
    
    # Create dataloaders
    train_loader = DataLoader(
        train_dataset, 
        batch_size=Config.BATCH_SIZE, 
        shuffle=True, 
        num_workers=4
    )
    test_loader = DataLoader(
        test_dataset, 
        batch_size=Config.BATCH_SIZE, 
        shuffle=False, 
        num_workers=4
    )
    
    # 2. Model Initialization
    print("\n2. Initializing CCAL model...")
    model = CCAL_Model()
    print(f"Total parameters: {sum(p.numel() for p in model.parameters()):,}")
    print(f"Trainable parameters: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}")
    
    # 3. Training
    print("\n3. Starting training...")
    trainer = Trainer(model)
    final_metrics = trainer.train(train_loader, test_loader)
    
    # 4. Evaluation
    print("\n4. Final Evaluation Results:")
    print("-" * 50)
    print(f"Accuracy: {final_metrics['accuracy']*100:.2f}%")
    print(f"Precision: {final_metrics['precision']*100:.2f}%")
    print(f"Recall: {final_metrics['recall']*100:.2f}%")
    print(f"F1-Score: {final_metrics['f1']*100:.2f}%")
    print(f"AUC: {final_metrics['auc']*100:.2f}%")
    
    # Per-class metrics
    print("\nPer-class metrics:")
    for class_name in Config.CLASS_NAMES:
        print(f"\n{class_name.capitalize()}:")
        print(f"  Precision: {final_metrics[f'{class_name}_precision']*100:.2f}%")
        print(f"  Recall: {final_metrics[f'{class_name}_recall']*100:.2f}%")
        print(f"  F1-Score: {final_metrics[f'{class_name}_f1']*100:.2f}%")
    
    # 5. Visualizations
    print("\n5. Generating visualizations...")
    plot_training_history(trainer.history)
    plot_confusion_matrix(final_metrics['confusion_matrix'], Config.CLASS_NAMES)
    
    # 6. Optional: K-fold Cross-Validation
    perform_cv = input("\nPerform 5-fold cross-validation? (y/n): ")
    if perform_cv.lower() == 'y':
        print("\n6. Performing k-fold cross-validation...")
        full_dataset = LungCancerDataset(
            Config.DATA_PATH, 
            transform=train_transform, 
            train=True, 
            train_split=1.0
        )
        fold_results, avg_metrics = k_fold_cross_validation(full_dataset)
        
        print("\nCross-validation results:")
        print("-" * 50)
        for key in ['accuracy', 'precision', 'recall', 'f1', 'auc']:
            print(f"{key.capitalize()}: {avg_metrics[key]*100:.2f}% (±{avg_metrics[f'{key}_std']*100:.2f}%)")
    
    print("\n" + "=" * 60)
    print("Training completed successfully!")
    print("=" * 60)

if __name__ == "__main__":
    main()
```
