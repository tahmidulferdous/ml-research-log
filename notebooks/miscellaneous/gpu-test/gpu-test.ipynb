```python
import tensorflow as tf
import time

def tpu_flops_test():
    # Enable TPU
    resolver = tf.distribute.cluster_resolver.TPUClusterResolver.connect()
    strategy = tf.distribute.TPUStrategy(resolver)

    # Create a simple computation graph
    with strategy.scope():
        a = tf.random.uniform([1000, 1000], dtype=tf.float32)
        b = tf.random.uniform([1000, 1000], dtype=tf.float32)

        start_time = time.time()
        for _ in range(1000):  # Repeat to simulate workload
            c = tf.matmul(a, b)
        end_time = time.time()

        elapsed_time = end_time - start_time
        flops_per_matmul = 2 * (1000**3)  # FLOPs for matrix multiplication
        total_flops = flops_per_matmul * 1000
        tpu_flops = total_flops / elapsed_time

    print(f"TPU FLOPs: {tpu_flops / 1e12:.2f} TFLOPs, Time: {elapsed_time:.4f} seconds")

if __name__ == "__main__":
    tpu_flops_test()
```
