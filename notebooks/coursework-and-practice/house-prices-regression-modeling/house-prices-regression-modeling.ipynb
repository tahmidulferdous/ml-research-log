```python
def get_unique_values(original_data, num_unique_values = 10): 
    '''
    This function return a DataFrame of variables names and their unique values.
    original_data = main dataframe 
    num_unique_values = number of unique values not more than given
    '''
    # Create a list to store the unique values
    unique_values_list = []

    # Iterate through the columns of the original dataset
    for column_name in original_data.columns:
        
        if len(original_data[column_name].unique()) <= num_unique_values:
            unique_values = original_data[column_name].unique() 

            # Convert unique values to a comma-separated string
            unique_values_str = ', '.join(map(str, unique_values))

            unique_values_list.append({'Variable': column_name, 'Unique_Values': unique_values_str})

    # Create a new DataFrame from the list of dictionaries
    unique_values_df = pd.DataFrame(unique_values_list)
    return unique_values_df
var_ana = get_unique_values(train)
var_ana
```

```python
def correlation_vizualization(df,y, num_cols = 4 ):    # Adjust this based on the number of columns you want
    num = df.select_dtypes(exclude='object')
    num_feature = list(num.columns)
    num_rows = -(-len(num.columns) // num_cols)  # Ceiling division to determine the number of rows

    fig, axes = plt.subplots(num_rows, num_cols, figsize=(15, num_rows * 5))

    # Flatten the 2D array of axes to simplify indexing
    axes = axes.flatten()

    # Plot scatter plots for each numerical feature
    for i, feature in enumerate(num.columns):
        ax = axes[i]
        ax.scatter(num[feature], num[y])
        ax.set_xlabel(feature)
        ax.set_ylabel(y)
        correlation_coefficient = num[feature].corr(num[y])
        ax.set_title(f'Scatter plot for {feature}\nCorrelation: {correlation_coefficient:.2f}')

    # Remove empty subplots
    for i in range(len(num.columns), len(axes)):
        fig.delaxes(axes[i])

    # Adjust layout to prevent clipping
    plt.tight_layout()
    plt.show()
```

```python
correlation_vizualization(train,'SalePrice')
```
