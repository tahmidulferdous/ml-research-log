```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import ipywidgets as widgets
from ipywidgets import interact

# Generate random data for demonstration
np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)

# Initial values for m and b
initial_m = 2
initial_b = 3

# Function to calculate total squared error
def calculate_squared_error(X, y, m, b):
    predictions = m * X + b
    squared_error = np.sum((predictions - y)**2)
    return squared_error

# Function to plot the regression line and total squared error
def plot_regression_line(m, b):
    # Create a DataFrame for Seaborn
    data = {'X': X.flatten(), 'y': y.flatten()}
    df = pd.DataFrame(data)

    # Scatter plot for data points
    sns.scatterplot(x='X', y='y', data=df, label='Data points')

    # Line plot for regression line
    sns.lineplot(x='X', y=m * X.flatten() + b, color='red', label=f'Regression Line: y = {m}*x + {b}')

    # Calculate and display total squared error
    total_squared_error = calculate_squared_error(X, y, m, b)
    plt.text(1.5, 25, f'Total Squared Error: {total_squared_error:.2f}', style='italic',
             bbox={'facecolor': 'yellow', 'alpha': 0.5, 'pad': 10})

    plt.title('Linear Regression Visualization')
    plt.xlabel('X')
    plt.ylabel('y')
    plt.legend()
    plt.grid(True)
    plt.show()

# Create interactive sliders for m and b
m_slider = widgets.FloatSlider(value=initial_m, min=-5, max=5, step=0.1, description='m')
b_slider = widgets.FloatSlider(value=initial_b, min=-5, max=5, step=0.1, description='b')

# Use the interactive decorator to connect the sliders to the plot function
interact(plot_regression_line, m=m_slider, b=b_slider)
```
